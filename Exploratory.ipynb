```python
import matplotlib.pyplot as plt
import pandas as pd
from ipywidgets import widgets
```

```python
widget = widgets.Dropdown(
    options=[
        "matplotlib",
        "hvplot",
        "holoviews",
        "pandas_bokeh",
        "plotly",
        "altair",
    ],
    value="matplotlib",
    description="Select Backend: ",
    disabled=False,
)
```

```python
from IPython.display import display


def select_backend(backend=widget):
    display(widget)
    pd.set_option("plotting.backend", widget.value)
```

```python
select_backend()
```

```python
TEMP = pd.read_parquet("temp.parquet")
PRECIPITATION = pd.read_parquet("precipitation.parquet")
```

```python
# TEMP.query("fips == 12086").truncate(before="11/1993")[
TEMP.query("fips == 12086")[["avg_temp", "max_temp", "min_temp"]].plot(
    title="Temperature: Miami-Dade"
)
```
