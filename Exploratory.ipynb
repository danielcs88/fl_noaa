```python
import matplotlib.pyplot as plt
import pandas as pd
from ipywidgets import widgets
```

```python
widget = widgets.Dropdown(
    options=[
        "matplotlib",
        "hvplot",
        "holoviews",
        "pandas_bokeh",
        "plotly",
        "altair",
    ],
    value="matplotlib",
    description="Select Backend: ",
    disabled=False,
)
```

```python
from IPython.display import display


def select_backend(backend=widget):
    display(widget)
    pd.set_option("plotting.backend", widget.value)
```

```python
select_backend()
```

```python
TEMP = pd.read_parquet("temp.parquet")
PRECIPITATION = pd.read_parquet("precipitation.parquet")
```

```python
TEMP.query("fips == 12086").truncate(before="11/1993")[
    ["avg_temp", "max_temp", "min_temp"]
].plot(title="Temperature: Miami-Dade")
```

```python
TEMP.query("fips == 12086").truncate(before="11/1993")[
    ["avg_temp", "max_temp", "min_temp"]
].pct_change(12).plot(title="Temperature: Miami-Dade\nPercent Changes YoY")
```

```python
PRECIPITATION.query("fips == 12086").truncate(before="11/1993")["precipitation"].plot(
    title="Precipitation: Miami-Dade"
)
```

```python
PRECIPITATION.query("fips == 12086").truncate(before="11/1993")[
    "precipitation"
].pct_change(12).plot(title="Precipitation: Miami-Dade\nPercent Changes YoY")
```
